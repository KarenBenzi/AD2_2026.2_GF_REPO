# Demostración del sistema experto de asignación de aulas

Ejecutar desde la raíz del repositorio. Los datos son de ejemplo. Este notebook muestra entradas, salidas y traza resumida.

```
from reglas import RegistroOcupaciones, evaluar_motor
from datos import OCUPACIONES_INICIALES

registro = RegistroOcupaciones(list(OCUPACIONES_INICIALES))
```

```
casos = [
    {"id": "CP-demo-01", "estudiantes": 18, "dia": "lunes", "franja": "06:00–08:00", "proyector": False, "pcs": False},
    {"id": "CP-demo-02", "estudiantes": 20, "dia": "lunes", "franja": "08:00–10:00", "proyector": False, "pcs": False},
    {"id": "CP-demo-03", "estudiantes": 15, "dia": "jueves", "franja": "06:00–08:00", "proyector": False, "pcs": True},
]

for caso in casos:
    motor, validas, motivos = evaluar_motor(
        caso["estudiantes"], caso["dia"], caso["franja"],
        caso["proyector"], caso["pcs"], registro
    )
    print(f"\n{caso['id']}: {caso['estudiantes']} estudiantes, {caso['dia']} {caso['franja']}")
    print("Aula recomendada:", validas[0] if validas else "No hay aula válida")
    if not validas:
        print("Motivos:", motivos)
    print("Reglas disparadas:", ", ".join(dict.fromkeys(p.split(":", 1)[0] for p in motor.traza)))
```
